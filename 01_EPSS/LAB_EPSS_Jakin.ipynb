# NVD data

````python
base_url = "https://services.nvd.nist.gov/rest/json/cves/2.0"
date_start_NVD = '2025-09-01T00:00:00.000Z' # Do NOT change these dates
date_end_NVD   = '2025-10-01T00:00:00.000Z' # Do NOT change these dates
start_index = 0
results_per_page = 1000
total_results = 1 

all_cves = []

while start_index < total_results:
    params = {
        "pubStartDate": date_start_NVD,
        "pubEndDate": date_end_NVD,
        "resultsPerPage": results_per_page,
        "startIndex": start_index,
        "noRejected": ""
    }
    response = requests.get(base_url, params=params, timeout=6)
    if response.status_code != 200:
        print("Error:", response.status_code)
        break

    data = response.json()
    total_results = data.get("totalResults", 0)

    all_cves.extend(data.get("vulnerabilities", []))

    start_index += results_per_page
    print(start_index)
````

````python
# normalize and preprocess data
df = pd.json_normalize(all_cves, record_path=None, sep='.', max_level=None)
df = preprocess_NVD_data(df)

# remove vulnerabilities marked as "reject" or "reserved"
df = df[(df['cve.vulnStatus']!='Reserved') & (df['cve.vulnStatus']!='Reject')]

# merge NVD and EPSS data
df = df.merge(epss_current, left_on="cve.id", right_on="cve", how="left")
````

## CVSS Base Score Analysis

Let's examine the distribution and descriptive statistics of the CVSS base scores to understand the severity levels of the vulnerabilities in our dataset.

````python
# Analyze CVSS base score distribution
cvss_stats = df['cvss_baseScore'].describe()
print("CVSS Base Score Statistics:")
print(cvss_stats)

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='cvss_baseScore', bins=20)
plt.title('Distribution of CVSS Base Scores')
plt.xlabel('CVSS Base Score')
plt.ylabel('Count')
plt.show()
````



Looking at the statistics and the chart, we can see how dangerous these vulnerabilities are:

- **Mostly High Scores**: The chart leans heavily to the right. This means most vulnerabilities are rated as **High (7.0-8.9)** or **Critical (9.0-10.0)**.
  
- **Average Severity**: The average score is in the "Medium" range getting towards to the "High" range, which tells us that the typical vulnerability in this dataset is a serious risk.
  
- **Why this matters**: We see very few "Low" issues. This suggests that researchers and companies might be prioritizing the most dangerous bugs, or that the current security landscape is full of high-impact problems.

````python

def cvss_categorization(score):
    if pd.isna(score):
        return 'Unknown/NA'
    elif score < 4.0:
        return 'Low'
    elif 4.0 <= score < 7.0:
        return 'Medium'
    elif 7.0 <= score < 9.0:
        return 'High'
    else:
        return 'Critical'

df['cvss_category'] = df['cvss_baseScore'].apply(cvss_categorization)

# Calculate percentages of each CVSS category
total_vulns = len(df[df['cvss_baseScore'].notna()])
category_percentages = (df['cvss_category'].value_counts() / total_vulns * 100).round(2)

# Exclude the 'Unknown/NA' label and order categories by severity
desired_order = ['Low', 'Medium', 'High', 'Critical']
# Keep only the desired categories in that order and drop missing ones
category_percentages = category_percentages.reindex(desired_order).dropna()

# Create a pie chart
plt.figure(figsize=(12, 6))

# Subplot 1: Pie chart
plt.subplot(1, 2, 1)
palette = {'Low':'green', 'Medium':'yellow', 'High':'orange', 'Critical':'red'}
colors = [palette[c] for c in category_percentages.index]
plt.pie(category_percentages, labels=category_percentages.index, colors=colors, autopct='%1.1f%%')
# Add a legend describing the numeric CVSS ranges for each category

range_labels = {'Low':'0.0–3.9', 'Medium':'4.0–6.9', 'High':'7.0–8.9', 'Critical':'9.0–10.0'}
handles = [Patch(facecolor=palette[c], label=f"{c} ({range_labels[c]})") for c in category_percentages.index]
plt.legend(handles=handles, title='CVSS range', loc='center left', bbox_to_anchor=(1, 0.5))
plt.title('Distribution of CVSS Categories')

# Subplot 2: Box plot by category
plt.subplot(1, 2, 2)
#sns.boxplot(data=df, x='cvss_category', y='cvss_baseScore', order=desired_order, 
          # palette=colors)
# Restrict boxplot to same categories and use same order/colors
sns.boxplot(data=df[df['cvss_category'].isin(category_percentages.index)], 
            x='cvss_category', y='cvss_baseScore', 
            order=category_percentages.index, palette=colors)
plt.xticks(rotation=45)
plt.title('CVSS Score Distribution by Category')
plt.xlabel('CVSS Category')
plt.ylabel('CVSS Base Score')

plt.tight_layout()
plt.show()

# Print summary statistics by category (exclude Unknown/NA and keep desired order)
summary = df[df['cvss_category'].isin(desired_order)].groupby('cvss_category')['cvss_baseScore'].describe().reindex(desired_order)
print("\nSummary Statistics by CVSS Category:")
print(summary)
````



These charts help us understand the severity labels better:

- **Most bugs are Medium or High**: The pie chart shows that **Medium** and **High** are the most common labels. **Critical** issues are fewer but still a big chunk of the pie. **Low** severity bugs are very rare.
  
- **Clear Groups**: The box plot shows that the scores fit neatly into their categories. You can see that "Critical" bugs are always at the very top (9.0 to 10.0).
  
- **Connecting to the previous chart**: The histogram showed us the scores were generally high. This breakdown confirms it: if you add up High and Critical, they make up a large part of the total. This means security teams have a lot of serious work to do, as they can't just ignore the "Medium" ones either.

## CVSS vs EPSS Correlation

Let's create a scatter plot to visualize the relationship between CVSS base scores and EPSS scores to understand if there's any correlation between vulnerability severity and exploitation probability.

````python
filtered_data = df.dropna(subset=['cvss_baseScore', 'epss'])

# Creating scatter plot of CVSS vs EPSS
plt.figure(figsize=(10, 6))
sns.scatterplot(data=filtered_data, 
                x='cvss_baseScore',
                y='epss',
                alpha=0.5,
                color='red')
plt.title('CVSS Base Score vs EPSS')
plt.xlabel('CVSS Base Score')
plt.ylabel('EPSS Score')

# Calculating correlation
correlation = filtered_data['cvss_baseScore'].corr(filtered_data['epss'])
plt.text(0.05, 0.95, f'Correlation: {correlation:.3f}', 
         transform=plt.gca().transAxes, 
         bbox=dict(facecolor='white', alpha=0.8))
plt.show()
````

#### CVSS vs EPSS Analysis

- **Near-Zero Correlation**: The scatter plot shows almost no statistical connection between CVSS and EPSS, with a correlation of just 0.060.
  
- **Flat Line**: Visually, there is a dense horizontal line of red dots along the very bottom of the chart (EPSS ≈ 0). This line stretches across every severity level, from low (2.0) all the way to critical (10.0).
  
- **Severity ≠ Likelihood**: This pattern confirms that a "critical" CVSS score does not automatically imply a high chance of exploitation.
  
- **Outliers**: While the few most dangerous outliers in the top right do have high severity scores, the vast majority of high-severity vulnerabilities in the dataset have almost zero probability of being weaponized.
  
- **Important aspect**: The plot suggests that severity alone does not determine the likelihood of exploitation, hence there might be other factors such as target popularity and attack complexity that play a crucial role in the prediction.
 

## Most Frequent Vendors

In the following section we try to extract the vendor identities from "vulnerable_cpes" and visualize the top 20 most prevalent vendors in the dataset.

````python
vendors = []
for cpes in df['vulnerable_cpes'].dropna():
    for cpe in cpes:
        parts = cpe.split(':')
        if len(parts) >= 4:     # Extracting vendor from CPE string (4th field after splitting by ':')
            vendor = parts[3]
            vendors.append(vendor)

vendor_counts = Counter(vendors).most_common(20)
vendor_df = pd.DataFrame(vendor_counts, columns=['Vendor', 'Count'])

# Set a clean background style
sns.set_style("whitegrid")

plt.figure(figsize=(12, 8))



ax = sns.barplot(
    data=vendor_df, 
    x='Count', 
    y='Vendor', 
    palette='viridis', 
    hue='Vendor',  
    legend=False   
)

# Adding exact numbers to the end of the bars
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=10, color='black')

plt.xlabel('Number of CVEs', fontsize=12)
plt.ylabel('Vendor', fontsize=12)
plt.title('Top 20 Most Frequent Vendors', fontsize=15, fontweight='bold')

# Removing the box border for a cleaner look
sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
````

- Product Vendors

## Feature Engineering

I am about to create additional and more meaninfgul feature to analyze potential exploitability trends.

- <b> has_github_ref:</b>
  - Set to TRUE if references contain a GitHub link -> PoC code indicator

- <b>is_dangerous_cwe:</b>
  - Storing the information whether a CVE has historically dangerous CWE

- <b>is_popular_vendor:</b>
  - Storing info whether a CVE is encountered in a popular vendor's product.

- <b>days_to_modify:</b>
  - Number of days between publication and last modification

````python
# CONTAINS_GITHUB_REF
def contains_github(refs):
    return 'github' in str(refs).lower()

low_epss_df.loc[:, 'has_github_ref'] = low_epss_df['cve.references'].apply(contains_github)



# DANGEROUS_CWES
dangerous_cwes = [
    'CWE-78',   # OS Command Injection
    'CWE-79',   # Cross-site Scripting (XSS)
    'CWE-89',   # SQL Injection
    'CWE-22',   # Path Traversal
    'CWE-434',  # Unrestricted Upload
    'CWE-94',   # Code Injection
    'CWE-77',   # Command Injection
    'CWE-502',  # Deserialization
    'CWE-20',   # Improper Input Validation
    'CWE-119',  # Buffer Overflow
    'CWE-787',  # Out-of-bounds Write
    'CWE-121',  # Stack-based Buffer Overflow
    'CWE-74',   # Injection
] # Compare the most dangerous CWEs with the most frequent CWEs from previous analysis


def has_dangerous_cwe(cwe_list):
    if not cwe_list:
        return False
    for cwe in cwe_list:
        if cwe in dangerous_cwes:
            return True
    return False

low_epss_df.loc[:, 'has_dangerous_cwe'] = low_epss_df['cwe_list'].apply(has_dangerous_cwe)






# POPULAR_VENDORS
# First we need to extract the vendor from the CPE strings
def extract_vendor(cpes):
    if not cpes or not isinstance(cpes, list):
        return 'unknown'
    
    parts = cpes[0].split(':')
    if len(parts) >= 4:
        return parts[3]
    return 'unknown'

low_epss_df.loc[:, 'vendor'] = low_epss_df['vulnerable_cpes'].apply(extract_vendor)

# List of popular vendors to focus on -> Some correlations with the most frequent vendors!
popular_vendors = [
    'dell', 'cisco', 'microsoft', 'qualcomm',
    'google', 'apple', 'adobe', 'ivanti',
    'ibm', 'gitlab', 'samsung', 'hp', 'linux'
]

def is_popular_vendor(vendor):
    # 1. Check if it's actually a string first
    if not isinstance(vendor, str):
        return False
        
    # 2. Check the specific conditions
    if vendor == 'unknown':
        return False
        
    return vendor.lower() in popular_vendors

low_epss_df.loc[:, 'is_popular_vendor'] = low_epss_df['vendor'].apply(is_popular_vendor)




# DAYS_TO_MODIFY
low_epss_df['days_to_modify'] = (
    pd.to_datetime(low_epss_df['cve.lastModified']) -
    pd.to_datetime(low_epss_df['cve.published'])
).dt.days.fillna(0)


low_epss_df[['cve.id', 'epss', 'percentile', 'num_references', 'has_github_ref', 
'has_dangerous_cwe', 'is_popular_vendor', 'days_to_modify']].head(20)

#low_epss_df[['days_to_modify']].sort_values('days_to_modify', ascending=False).head(20)

#print(f"OK CVEs with dangerous CWEs: {low_epss_df['has_dangerous_cwe'].sum()}/{len(low_epss_df)}")
#print(f"OK CVEs from popular vendors: {low_epss_df['is_popular_vendor'].sum()}/{len(low_epss_df)}")
````

The application collected daily probability scores from **October 1, 2025** to present. The objective was to determine if risk scores remained stable or exhibited "velocity" (rapid growth).

![EPSS results on 11/12/2025](epss_streamlit_result.png)
*Figure 1: Analysis of EPSS scores over a 2-month period.*

The visual data confirms that risk is not static. While the majority of vulnerabilities remained flat, specific outliers demonstrated clear upward trends:

* **🔴 CVE-2025-11031 (Red Line):** This had the biggest change. It started low (**0.05%**) and tripled at **0.17%**.
* **🟣 CVE-2025-9789 (Purple Line):** Doubled in score between October and November before staying flat again.

### Takeaways

**1. The "Step Function" of Risk**

The trend lines do not rise smoothly; they jump in discrete **steps**. This suggests the score reacts to specific trigger events, such as a hacker attempting an exploit or the publication of new threat intelligence online.

**2. Relative Growth vs. Absolute Score**

Although the absolute probabilities remain low (<1%), there is some relative growth. The Red line grew by roughly **300%** in two months. In a risk-based approach, this high velocity is an early warning sign that the vulnerability is transitioning into an active threat.

 Monitoring the *rate of change* (trend) is just as critical as analyzing the static score itself.
